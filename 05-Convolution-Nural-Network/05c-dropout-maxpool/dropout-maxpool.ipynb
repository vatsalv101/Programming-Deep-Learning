## Dropout
---

Dropout randomly sets activations to zero during training. This reduces co-adaptation between features and often improves generalization.

In this exercise, we use inverted dropout. If an activation $x_i$ stays active with probability $1-p$, it is scaled by $\frac{1}{1-p}$ during training:

$$
y_i = x_i \cdot m_i,
\qquad
m_i \in \left\{0, \frac{1}{1-p}\right\}.
$$

With inverted dropout, evaluation mode is simple: the layer returns the input unchanged.

## Max-Pooling
---

Max-pooling downsamples (reduces size of) a feature map (output of a conv. layer) by taking the maximum value inside each local window.

For an input tensor $x \in \mathbb{R}^{N \times C \times H \times W}$, a max-pooling layer is applied independently for every batch element and channel.
