---
A convolution layer takes an input tensor $x \in \mathbb{R}^{N \times C_{in} \times H \times W}$
and applies filters $W \in \mathbb{R}^{C_{out} \times C_{in} \times K_h \times K_w}$
with bias $b \in \mathbb{R}^{C_{out}}$.

The output tensor has shape
$y \in \mathbb{R}^{N \times C_{out} \times H_{out} \times W_{out}}$
with

$$
H_{out} = \left\lfloor \frac{H + 2p - K_h}{s} \right\rfloor + 1
\qquad\text{and}\qquad
W_{out} = \left\lfloor \frac{W + 2p - K_w}{s} \right\rfloor + 1,
$$

where $s$ is the stride and $p$ is the padding.

For one output value, the convolution computes

$$
y_{n,c,i,j} =
\sum_{d=1}^{C_{in}}
\sum_{u=1}^{K_h}
\sum_{v=1}^{K_w}
W_{c,d,u,v} \cdot x^{pad}_{n,d,i \cdot s + u, j \cdot s + v}
+ b_c.
$$

In the backward pass, every scalar gradient
$\frac{\partial L}{\partial y_{n,c,i,j}}$
contributes to three different quantities:

- the bias gradient of channel $c$,
- the gradient of the filter $W_c$,
- and the gradient of the corresponding input patch in the padded input.

---
In this notebook, the provided forward pass returns both the output and a cache:
```python
out, cache = conv2d_forward_4_loops(x, weight, bias, stride, padding)
```

Your backward implementation will use this cache and should return
```python
dx, dweight, db = conv2d_backward_4_loops(dout, cache)
```
