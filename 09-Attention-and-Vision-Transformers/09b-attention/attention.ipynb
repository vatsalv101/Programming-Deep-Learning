## Theory: Attention Calculation
---

Attention turns a sequence of vectors into a new sequence by letting every position compute a weighted average over all other positions. A sequence of $n$ different $d_k$-dimensional key and query vectors is used to calculate attention weights for all $n$ $d_v$-dimensional value vectors, which are then summed up to produce the output sequence. More formally using Matrix notation: 

Given queries $Q \in \mathbb{R}^{n \times d_k}$, keys $K \in \mathbb{R}^{n \times d_k}$, and values $V \in \mathbb{R}^{n \times d_v}$, scaled dot-product attention is

$$
\mathrm{Attention}(Q, K, V) = \mathrm{softmax}\!\left(\frac{QK^\top}{\sqrt{d_k}}\right)V.
$$

The softmax is taken over the key dimension, so each row of the attention matrix sums to 1.

For **self-attention**, the same input sequence $X \in \mathbb{R}^{n \times d}$ is used to build all three components:

$$
Q = XW_Q, \qquad K = XW_K, \qquad V = XW_V,
$$

where $W_Q, W_K \in \mathbb{R}^{d \times d_k}$ and $W_V \in \mathbb{R}^{d \times d_v}$ are learned projections and $d$ is the original token dimension.

For **multi-head self-attention**, this computation is repeated $h$ times with separate projection matrices:

$$
\mathrm{head}_i = \mathrm{Attention}(XW_Q^{(i)}, XW_K^{(i)}, XW_V^{(i)}),
$$

where $i \in {1,\dots, h}$. The head outputs are concatenated and projected back to the model dimension:

$$
\mathrm{MultiHeadAttention}(X) = [\mathrm{head}_1; \dots; \mathrm{head}_h]W_O,
$$

where $W_O \in \mathbb{R}^{h\cdot d_v \times d}$ is a learnable output projection. Usually $d_k = d_v = d/h$, so the heads split the representation into smaller subspaces while keeping the total output dimension unchanged.

This gives each head a chance to focus on different relationships in the sequence, while the final output remains a full sequence of contextualized embeddings.
