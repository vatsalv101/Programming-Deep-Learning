## Python Setup

To run a jupyter notebook, you first need to install python and setup a python environment. In this environment, you can install external packages. To isolate packages in multiple project, it is recommended to create a virtual environment for each project. For managing these environment and python packages, we recommend using [uv](https://docs.astral.sh/uv/). We recommend to structure your project like this:

```
programming-deep-learning
├── pyproject.toml
└── exercises
    ├── exercise01
    ├── exercise02
    └── ...
```

This means that the pyproject.toml should be on the top level of your project structure! For a complete explanation on how to structure a project, check the [uv projects guide](https://docs.astral.sh/uv/guides/projects/).

### uv setup

0. Install [uv](https://docs.astral.sh/uv/).
1. In your project folder(`programming-deep-learning`), run `uv sync` in your terminal to install the packages specified in the `pyproject.toml` file we provided. This will install the packages in `.venv` and create a `uv.lock` file. While the project file provides constraints for the packages (e.g. `numpy>=2.4`), the lockfile shows which version was actually installed (e.g. `version==2.4.1`) along with other information. The lockfile should not be modified manually, if you want to modify the packages use either the project file or run `uv add packagename`.
2. You may need to select the uv interpreter in your editor.
    * Pycharm: Bottom right _&lt;No interpreter&gt;_ -> _Add New Interpreter_ -> _Add Local Interpreter_ -> The created uv interpreter should be preselected.
    * VS Code: Open a jupyter notebook (like this file). Top right _Select Kernel_ -> _Python Environment_ -> _programming-deep-leraning_
3. You can now run python scripts with `uv run path/to/file.py`. If your installed packages do not comply with the `pyproject.toml`, uv will sync them automatically before running.

### Pytorch setup

For this lecture, we will work with pytorch as the central component for scientific computing with a focus on deep learning. One on its main features is the use of GPUs to accelerate computing, hence pytorch offers specific versions for different GPUs. Without the massiv parallelization of GPUs, training deep neural network would not be possible in any feasible time. We try to provide you exercises that are small enough to be possible to solve with just the CPU, but if you have the option, use a GPU.  Currently, the configuration in the `pyproject.toml` is set to CPU only. This guide shows how to setup pytorch with an Nvidia GPU.

1. Install the [cuda toolkit](https://developer.nvidia.com/cuda-downloads). Needs to be same version as in step (3) e.g. version 13.0
2. Run `nvidia-smi` and check which cuda version is installed. This version is the upper bound of the pytorch version you should install.
3. Set the correct pytorch cuda version for pytorch in the `pyproject.toml` in the `[tool.uv.sources]` section. There are currently sources prepared for `cpu`, `cu128` and `cu130`.
If you now run `uv sync`, the correct cuda version will be installed. If you have an AMD or Intel GPU or need another cuda version, you have to add a new source.
