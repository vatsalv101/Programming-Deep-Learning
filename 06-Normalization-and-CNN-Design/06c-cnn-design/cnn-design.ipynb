## PyTorch devices
---

In PyTorch, a **device** is an object that represents the hardware on which tensors are stored and computations are performed.

The default device is `device = torch.device("cpu")`, which means that tensors are stored in the main memory (RAM) and computations are performed on the CPU.

For faster training, CUDA-capable GPUs can be used.  
Then the device is `device = torch.device("cuda")`, tensors are stored in the GPU memory, and computations are performed on the GPU.

Both tensors and instances of `nn.Module` (like our `SimpleCNN`) can be moved to a device using the `.to(device)` method:
- `tensor = tensor.to(device)` returns a new tensor on the specified device. If the tensor is already on that device, it may return the same tensor.
- `model = SimpleCNN(); model = model.to(device)` moves all model parameters and buffers to the specified device and returns the model itself.
